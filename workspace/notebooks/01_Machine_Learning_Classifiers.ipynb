# Machine Learning: Classification with Scikit-Learn 🤖

In this notebook, we train a **Random Forest Classifier** to predict whether a student will pass the exam based on study hours, attendance percentage, and previous score.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import matplotlib.pyplot as plt

# 1. Load dataset
df = pd.read_csv('workspace/data/students.csv')
features = ['study_hours', 'attendance_pct', 'previous_score']
X = df[features]
y = df['passed_exam']

# 2. Train / Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# 3. Fit Random Forest Model
model = RandomForestClassifier(n_estimators=50, random_state=42)
model.fit(X_train, y_train)

# 4. Evaluate
y_pred = model.predict(X_test)
acc = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {acc * 100:.1f}%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# 5. Plot Feature Importances
plt.figure(figsize=(6, 3))
plt.barh(features, model.feature_importances_, color='#10b981')
plt.title('Random Forest Feature Importance')
plt.xlabel('Relative Importance')
plt.show()
```
