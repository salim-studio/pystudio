# Econometrics & Symbolic Math 📐

Here we estimate an **Ordinary Least Squares (OLS)** linear regression model using `statsmodels` to quantify the elasticity of final grades with respect to study hours and attendance, and solve symbolic calculus equations with `sympy`.

```python
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf

df = pd.read_csv('workspace/data/students.csv')

# Ordinary Least Squares Regression
ols_model = smf.ols('final_grade ~ study_hours + attendance_pct', data=df).fit()
print(ols_model.summary().tables[1])
print(f"\nR-squared: {ols_model.rsquared:.4f}")
print(f"F-statistic p-value: {ols_model.f_pvalue:.4e}")
```
