### Interactive Scientific Calculation
Run the cell below to compute statistics with **NumPy** and plot with **Matplotlib**:

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 100 sample data points
x = np.linspace(0, 10, 100)
y = np.sin(x) + np.random.normal(0, 0.1, 100)

plt.figure(figsize=(7, 3.5))
plt.plot(x, y, label='Observed Signal', color='#3b82f6', lw=2)
plt.plot(x, np.sin(x), label='Ground Truth sin(x)', color='#ef4444', linestyle='--')
plt.title('NumPy + Matplotlib Live Plot')
plt.xlabel('Time (t)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print(f"Mean of y: {y.mean():.4f}, Std Dev: {y.std():.4f}")
```
